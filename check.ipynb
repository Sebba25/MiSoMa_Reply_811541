```python
import pandas as pd
import matplotlib.pyplot as plt


#load attivazioni cessazioni
df = pd.read_csv('attivazioniCessazioni.csv')

```

```python
# check column att ivazioni values
unique = df['attivazioni'].unique()
```

```python
# convert unique to numeric values, save non-numeric values in a separate variable
non_numeric = []
for value in unique:
    try:
        float(value)
    except ValueError:
        non_numeric.append(value)
```
